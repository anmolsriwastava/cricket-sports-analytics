```python
import numpy as np
import pandas as pd
from pathlib import Path
from sklearn.model_selection import GroupKFold
from sklearn.ensemble import HistGradientBoostingClassifier
```

```python
DATA = Path(r"C:\Users\Anmol Sriwastava\Desktop\RR_Analytics\data")
MIN_BAT_PRESSURE = 60
MIN_BAT_NORMAL = 120
MIN_BOWL_PRESSURE = 120
MIN_BOWL_NORMAL = 240
WICKET_VALUE = 11.5
K_BAT = 60
K_BOWL = 120
PRESSURE_Q = 0.75   # top 25% leverage

# ---------- load ----------
bbb = pd.read_csv(DATA / "ball_by_ball.csv")
matches = pd.read_csv(DATA / "matches.csv", parse_dates=["match_date"])
comps = pd.read_csv(DATA / "competitions.csv")

match_cols = [
    "match_id", "comp_id", "match_date", "stage", "knock_out",
    "revised_target", "first_inn_revised_overs", "second_inn_revised_overs",
    "venue_id", "win_team_id"
]
df = bbb.merge(matches[match_cols], on="match_id", how="left")
df = df.merge(comps[["comp_id", "match_format", "season"]], on="comp_id", how="left")

# normal innings only
df = df[df["innings"].isin([1, 2])].copy()
df = df.sort_values(["match_id", "innings", "innings_ball_num"]).reset_index(drop=True)

# ---------- approximate runs / legal balls ----------
df["extra_run"] = 0
df.loc[df["extras_type"].isin(["wide", "Overstep"]), "extra_run"] = 1
df["approx_runs"] = df["runs"].fillna(0) + df["extra_run"] + df["extras_2_runs"].fillna(0)

# byes/legbyes runs are missing in this dataset; leave as 0 and state limitation
df["legal_ball"] = ~df["extras_type"].isin(["wide", "Overstep"])

# bowler runs: bat runs + wide/no-ball penalty; byes/legbyes not conceded
df["bowler_runs"] = df["runs"].fillna(0) + df["extra_run"] + df["extras_2_runs"].fillna(0)
df.loc[df["extras_type"].isin(["byes", "legbyes"]), "bowler_runs"] = 0

df["striker_out"] = df["is_wicket"] & (df["dismissed_player_id"] == df["striker_id"])
non_bowler_wkts = ["RunOut", "Retire Out", "Obstructing the field", "Obstracting"]
df["bowler_wkt"] = df["is_wicket"] & (~df["dismissal_type"].isin(non_bowler_wkts))

# ---------- pre-ball state ----------
g = df.groupby(["match_id", "innings"], sort=False)
df["score_before"] = g["approx_runs"].cumsum() - df["approx_runs"]
df["wkts_before"] = g["is_wicket"].cumsum().astype(int) - df["is_wicket"].astype(int)
df["legal_before"] = g["legal_ball"].cumsum().astype(int) - df["legal_ball"].astype(int)

df["inn_total_balls"] = 120
df.loc[(df["innings"] == 1) & df["first_inn_revised_overs"].notna(), "inn_total_balls"] = (
    df["first_inn_revised_overs"] * 6
).round()
df.loc[(df["innings"] == 2) & df["second_inn_revised_overs"].notna(), "inn_total_balls"] = (
    df["second_inn_revised_overs"] * 6
).round()

df["balls_left_before"] = df["inn_total_balls"] - df["legal_before"]

first_total = (
    df[df["innings"] == 1]
    .groupby("match_id")["approx_runs"]
    .sum()
    .rename("first_total")
)
df = df.merge(first_total, on="match_id", how="left")
df["target"] = df["first_total"] + 1
df.loc[df["revised_target"].notna(), "target"] = df["revised_target"]
df.loc[df["innings"] == 1, "target"] = np.nan

# ---------- post-ball state ----------
df["score_after"] = df["score_before"] + df["approx_runs"]
df["wkts_after"] = df["wkts_before"] + df["is_wicket"].astype(int)
df["legal_after"] = df["legal_before"] + df["legal_ball"].astype(int)
df["balls_left_after"] = df["inn_total_balls"] - df["legal_after"]

df["curr_rate_before"] = np.where(df["legal_before"] > 0, df["score_before"] * 6 / df["legal_before"], 0)
df["curr_rate_after"] = np.where(df["legal_after"] > 0, df["score_after"] * 6 / df["legal_after"], 0)

df["runs_needed_before"] = df["target"] - df["score_before"]
df["runs_needed_after"] = df["target"] - df["score_after"]
df["req_rate_before"] = np.where(
    (df["innings"] == 2) & (df["balls_left_before"] > 0),
    df["runs_needed_before"] * 6 / df["balls_left_before"],
    np.nan,
)
df["req_rate_after"] = np.where(
    (df["innings"] == 2) & (df["balls_left_after"] > 0),
    df["runs_needed_after"] * 6 / df["balls_left_after"],
    np.nan,
)

df["is_t20i"] = (df["match_format"] == "T20I").astype(int)
df["bat_team_won"] = (df["win_team_id"] == df["batting_team_id"]).astype(int)
df.loc[df["win_team_id"].isna(), "bat_team_won"] = np.nan

# par score for innings 1: average score at same over by format/season
par = (
    df[df["innings"] == 1]
    .groupby(["match_format", "season", "overid"])["score_before"]
    .mean()
    .rename("par_score")
    .reset_index()
)
df = df.merge(par, on=["match_format", "season", "overid"], how="left")
df["score_minus_par"] = df["score_before"] - df["par_score"]

# ---------- cross-fitted WP ----------
def crossfit_wp_inn1(sub):
    sub = sub.dropna(subset=["bat_team_won"]).reset_index(drop=True)
    pre = pd.DataFrame({
        "score": sub["score_before"],
        "wkts": sub["wkts_before"],
        "balls_left": sub["balls_left_before"],
        "curr_rate": sub["curr_rate_before"],
        "overid": sub["overid"],
        "is_t20i": sub["is_t20i"],
        "score_minus_par": sub["score_minus_par"],
    }).replace([np.inf, -np.inf], np.nan)

    post = pd.DataFrame({
        "score": sub["score_after"],
        "wkts": sub["wkts_after"],
        "balls_left": sub["balls_left_after"],
        "curr_rate": sub["curr_rate_after"],
        "overid": sub["overid"],
        "is_t20i": sub["is_t20i"],
        "score_minus_par": sub["score_after"] - sub["par_score"],
    }).replace([np.inf, -np.inf], np.nan)

    y = sub["bat_team_won"].values
    groups = sub["match_id"].values
    wp_pre = np.full(len(sub), np.nan)
    wp_post = np.full(len(sub), np.nan)

    gkf = GroupKFold(n_splits=5)
    for tr, te in gkf.split(pre, y, groups):
        model = HistGradientBoostingClassifier(
            max_depth=3, learning_rate=0.05, max_iter=200, random_state=42
        )
        model.fit(pre.iloc[tr], y[tr])
        wp_pre[te] = model.predict_proba(pre.iloc[te])[:, 1]
        wp_post[te] = model.predict_proba(post.iloc[te])[:, 1]

    sub["wp_before"] = wp_pre
    sub["wp_after"] = wp_post
    sub["leverage"] = np.abs(sub["wp_after"] - sub["wp_before"])
    return sub


def crossfit_wp_inn2(sub):
    sub = sub.dropna(subset=["bat_team_won"]).reset_index(drop=True)
    pre = pd.DataFrame({
        "score": sub["score_before"],
        "wkts": sub["wkts_before"],
        "balls_left": sub["balls_left_before"],
        "runs_needed": sub["runs_needed_before"],
        "req_rate": sub["req_rate_before"],
        "curr_rate": sub["curr_rate_before"],
        "overid": sub["overid"],
        "is_t20i": sub["is_t20i"],
        "target": sub["target"],
    }).replace([np.inf, -np.inf], np.nan)

    post = pd.DataFrame({
        "score": sub["score_after"],
        "wkts": sub["wkts_after"],
        "balls_left": sub["balls_left_after"],
        "runs_needed": sub["runs_needed_after"],
        "req_rate": sub["req_rate_after"],
        "curr_rate": sub["curr_rate_after"],
        "overid": sub["overid"],
        "is_t20i": sub["is_t20i"],
        "target": sub["target"],
    }).replace([np.inf, -np.inf], np.nan)

    y = sub["bat_team_won"].values
    groups = sub["match_id"].values
    wp_pre = np.full(len(sub), np.nan)
    wp_post = np.full(len(sub), np.nan)

    gkf = GroupKFold(n_splits=5)
    for tr, te in gkf.split(pre, y, groups):
        model = HistGradientBoostingClassifier(
            max_depth=3, learning_rate=0.05, max_iter=200, random_state=42
        )
        model.fit(pre.iloc[tr], y[tr])
        wp_pre[te] = model.predict_proba(pre.iloc[te])[:, 1]
        wp_post[te] = model.predict_proba(post.iloc[te])[:, 1]

    sub["wp_before"] = wp_pre
    sub["wp_after"] = wp_post
    sub["leverage"] = np.abs(sub["wp_after"] - sub["wp_before"])
    return sub


inn1 = df[(df["innings"] == 1) & (df["balls_left_before"] > 0)].copy()
inn2 = df[(df["innings"] == 2) & (df["balls_left_before"] > 0)].copy()

inn1 = crossfit_wp_inn1(inn1)
inn2 = crossfit_wp_inn2(inn2)

df_wp = pd.concat([inn1, inn2], ignore_index=True)

# pressure label
threshold = df_wp["leverage"].quantile(PRESSURE_Q)
df_wp["is_pressure"] = df_wp["leverage"] >= threshold
print("leverage threshold:", threshold)
print(df_wp["is_pressure"].value_counts())

# ---------- player aggregation ----------
def agg_batter(df, min_pressure=MIN_BAT_PRESSURE, min_normal=MIN_BAT_NORMAL,
               wicket_value=WICKET_VALUE, k=K_BAT):
    b = df[df["legal_ball"]].copy()
    b["runs_bat"] = b["runs"].fillna(0)
    b["out"] = b["striker_out"].astype(int)

    def agg(sub):
        n = len(sub)
        return pd.Series({
            "balls": n,
            "runs": sub["runs_bat"].sum(),
            "outs": sub["out"].sum(),
            "rpb": sub["runs_bat"].sum() / n,
            "out_rate": sub["out"].sum() / n,
        })

    normal = b[~b["is_pressure"]].groupby(["striker_id", "striker_name"]).apply(agg).add_prefix("normal_")
    press = b[b["is_pressure"]].groupby(["striker_id", "striker_name"]).apply(agg).add_prefix("pressure_")

    out = press.join(normal, how="inner").reset_index()
    out = out[(out["pressure_balls"] >= min_pressure) & (out["normal_balls"] >= min_normal)].copy()

    out["delta_rpb"] = (
        (out["pressure_rpb"] - out["normal_rpb"])
        + wicket_value * (out["normal_out_rate"] - out["pressure_out_rate"])
    )
    out["shrunk"] = out["delta_rpb"] * out["pressure_balls"] / (out["pressure_balls"] + k)
    return out.sort_values("shrunk", ascending=False)


def agg_bowler(df, min_pressure=MIN_BOWL_PRESSURE, min_normal=MIN_BOWL_NORMAL,
               wicket_value=WICKET_VALUE, k=K_BOWL):
    b = df[df["legal_ball"]].copy()
    b["runs_bowl"] = b["bowler_runs"].fillna(0)
    b["wkt"] = b["bowler_wkt"].astype(int)

    def agg(sub):
        n = len(sub)
        return pd.Series({
            "balls": n,
            "runs": sub["runs_bowl"].sum(),
            "wkts": sub["wkt"].sum(),
            "rpb": sub["runs_bowl"].sum() / n,
            "wkt_rate": sub["wkt"].sum() / n,
        })

    normal = b[~b["is_pressure"]].groupby(["bowler_id", "bowler_name"]).apply(agg).add_prefix("normal_")
    press = b[b["is_pressure"]].groupby(["bowler_id", "bowler_name"]).apply(agg).add_prefix("pressure_")

    out = press.join(normal, how="inner").reset_index()
    out = out[(out["pressure_balls"] >= min_pressure) & (out["normal_balls"] >= min_normal)].copy()

    out["delta_rpb"] = (
        (out["normal_rpb"] - out["pressure_rpb"])
        + wicket_value * (out["pressure_wkt_rate"] - out["normal_wkt_rate"])
    )
    out["shrunk"] = out["delta_rpb"] * out["pressure_balls"] / (out["pressure_balls"] + k)
    return out.sort_values("shrunk", ascending=False)


batters = agg_batter(df_wp)
bowlers = agg_bowler(df_wp)

print("\nTOP BATTERS")
print(batters.head(10)[["striker_name", "shrunk", "pressure_balls", "normal_balls", "delta_rpb"]])
print("\nTOP BOWLERS")
print(bowlers.head(10)[["bowler_name", "shrunk", "pressure_balls", "normal_balls", "delta_rpb"]])

batters.to_csv("batter_pressure_rankings.csv", index=False)
bowlers.to_csv("bowler_pressure_rankings.csv", index=False)
```
